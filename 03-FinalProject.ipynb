### Epoch AI (Training-side compute proxies) — EDA

**What this dataset adds.** The Epoch AI Models dataset provides *training-side* cost proxies (e.g., Training compute (FLOP), Training chip-hours, and Training hardware) plus model scale (Parameters) for language-domain models. Leaderboards often report performance but not training compute, so Epoch helps us characterize how “cost” scales with model size.

**Note.** Here we load the cleaned Epoch dataset and show only the most relevant EDA figures for downstream cost–performance analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

EPOCH_PATH = "data/02-processed/epoch_ai_parameters.ndjson"
epoch_df = pd.read_json(EPOCH_PATH, lines=True)

print("Epoch cleaned shape:", epoch_df.shape)
print("Columns:", list(epoch_df.columns))
epoch_df.head()
```

#### Figure E1. Missingness in key Epoch compute columns
Before analyzing compute scaling, we quantify missingness for the cost-proxy columns we might use later.

```python
cols_to_check = [
    "Parameters",
    "Training compute (FLOP)",
    "Training chip-hours",
    "Training hardware",
    "Finetune compute (FLOP)",
    "Batch size",
]

# keep only columns that exist (prevents errors if a column name differs)
cols_to_check = [c for c in cols_to_check if c in epoch_df.columns]

missing_rate = epoch_df[cols_to_check].isna().mean().sort_values(ascending=False)

plt.figure(figsize=(8,4))
missing_rate.plot(kind="bar")
plt.title("Epoch AI: Missingness rate for key columns")
plt.xlabel("Column")
plt.ylabel("Fraction missing")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

missing_rate
```

**Interpretation.** Several training-cost fields have substantial missingness, especially `Training chip-hours` and `Batch size`. This matters because analyses using those columns rely on smaller subsets and could be biased if missingness is systematic. For downstream cost analysis, we report subset sample sizes and prioritize `Parameters` and `Training compute (FLOP)` when available.

#### Figure E2. Distribution of model size (Parameters)

Model size spans many orders of magnitude, so we visualize `Parameters` on a log10 scale.

```python
params = epoch_df["Parameters"].dropna()
params = params[params > 0]  # log10 requires positive values

plt.figure(figsize=(7,4))
plt.hist(np.log10(params), bins=40, edgecolor="white")
plt.title("Epoch AI: Distribution of model size (log10 Parameters)")
plt.xlabel("log10(Parameters)")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

**Interpretation.** Model sizes are highly skewed and cover multiple orders of magnitude, so log scaling is appropriate. This prevents the visualization from being dominated by extreme values and supports log–log modeling when relating size to compute.

#### Figure E3. Training compute vs parameter count (log–log)

Next, we examine whether larger models (more parameters) tend to require more training compute. Because both variables span many orders of magnitude, we analyze them in log–log space. This plot uses only rows where both values are present and positive.

```python
needed = ["Parameters", "Training compute (FLOP)"]
missing_needed = [c for c in needed if c not in epoch_df.columns]
if missing_needed:
    raise ValueError(f"Epoch dataset missing required columns: {missing_needed}")

sub = epoch_df.dropna(subset=needed).copy()
sub = sub[(sub["Parameters"] > 0) & (sub["Training compute (FLOP)"] > 0)]

x = np.log10(sub["Parameters"].values)
y = np.log10(sub["Training compute (FLOP)"].values)

r = np.corrcoef(x, y)[0, 1]
slope, intercept = np.polyfit(x, y, 1)

x_line = np.linspace(x.min(), x.max(), 200)
y_line = slope * x_line + intercept

plt.figure(figsize=(7,5))
plt.scatter(x, y, alpha=0.35, s=18)
plt.plot(x_line, y_line, linewidth=2, label="Linear fit (log–log)")
plt.title("Epoch AI: Training compute vs model size (log–log)")
plt.xlabel("log10(Parameters)")
plt.ylabel("log10(Training compute (FLOP))")
plt.legend()
plt.text(
    0.03, 0.97,
    f"n = {len(sub):,}\nr = {r:.3f}",
    transform=plt.gca().transAxes,
    ha="left", va="top",
    bbox={"boxstyle": "round", "facecolor": "white", "alpha": 0.9}
)
plt.tight_layout()
plt.show()

print(f"log–log correlation r = {r:.3f}")
print(f"log10(Training FLOPs) = {slope:.3f} * log10(Parameters) + {intercept:.3f}")
```

**Interpretation.** Among models with reported training compute, training FLOPs increase strongly with parameter count and appear approximately linear in log–log space. This supports using `Parameters` and `Training compute (FLOP)` as training-cost proxies, while noting that the result is conditional on the subset with reported compute.

**Connection to our main question.** Epoch provides the training-side “cost axis.” In later merged analyses (using benchmark/performance datasets handled elsewhere), we compare these cost proxies against performance to evaluate whether performance improvements exhibit diminishing returns as compute scales up.

### Epoch Analysis: Log–log regression of training compute on model size

EDA suggests a strong relationship between model size (`Parameters`) and training compute (`Training compute (FLOP)`) among models with reported compute values. Here we quantify that relationship with a log–log linear regression. This summarizes how quickly compute scales as models get larger, while explicitly restricting to rows where both values are present and positive.

```python
import statsmodels.api as sm

epoch_df = pd.read_json(EPOCH_PATH, lines=True)

sub = epoch_df.dropna(subset=["Parameters", "Training compute (FLOP)"]).copy()
sub = sub[(sub["Parameters"] > 0) & (sub["Training compute (FLOP)"] > 0)]

sub["log_params"] = np.log10(sub["Parameters"])
sub["log_train_flop"] = np.log10(sub["Training compute (FLOP)"])

X = sm.add_constant(sub["log_params"])
y = sub["log_train_flop"]

model = sm.OLS(y, X).fit()
print(model.summary())
```

**Interpretation.** The slope on `log_params` can be interpreted as an elasticity: it describes the expected change in log training compute for a one-unit increase in log parameter count. A strong fit (higher R²) supports the idea that parameter count is a meaningful proxy for training compute *within the subset that reports compute*. Because many models are missing compute values, this estimate should be interpreted as conditional on the reported-compute subset rather than a guarantee for all models in the full Epoch database.
